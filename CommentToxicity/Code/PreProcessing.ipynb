```python
import pandas as pd
import re
import emoji
from nltk import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

path = r"C:\Users\nisha\OneDrive\Desktop\Nisha\GUVI\MiniProject\CommentToxicity\Data\train.csv"
sw = stopwords.words('english')

df_train = pd.read_csv(path)
df_train.drop(['id'], axis=1,inplace=True)

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'http\S+|https\S+|www\S+','',text)
    text = re.sub(r"<.*?>",'',text)
    text = re.sub(r'@\w+|#\w+','',text)
    text = re.sub(r'(.)\1{2,}',r'\1\1',text)
    text = re.findall(r'[a-zA-Z]|\s',text)
    text = emoji.replace_emoji(text,'')
    text = re.sub(r'\s+',' ',text)
    tokens = word_tokenize(text)
    tokens = [token for token in tokens if token not in sw]
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(token) for token in tokens]
    return " ".join(tokens)

df_train['cleaned_comment'] = df_train['comment_text'].apply(clean_text)

X = df_train['cleaned_comment']
y = df_train[['toxic','severe_toxic','obscene','threat','insult','identity_hate']].values

vectorizer = TfidfVectorizer(max_features=100)
vectors = vectorizer.fit_transform(X)
vectors = pd.DataFrame(vectors.toarray(),columns = vectorizer.get_feature_names_out())

model = DecisionTreeClassifier()
model.fit(vectors,y)
y_pred = model.predict(vectors)

print(f"Accuracy: {accuracy_score(y,y_pred)}")
print(f"Precision: {precision_score(y,y_pred,average='micro')}")
print(f"Recall: {recall_score(y,y_pred,average='micro')}")
print(f"F1 Score: {f1_score(y,y_pred,average='micro')}")
```

```python
import pandas as pd
import re
import emoji
from nltk import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import tensorflow as tf

path = r"C:\Users\nisha\OneDrive\Desktop\Nisha\GUVI\MiniProject\CommentToxicity\Data\train.csv"
sw = stopwords.words('english')

df_train = pd.read_csv(path)
df_train.drop(['id'], axis=1,inplace=True)

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'http\S+|https\S+|www\S+','',text)
    text = re.sub(r"<.*?>",'',text)
    text = re.sub(r'@\w+|#\w+','',text)
    text = re.sub(r'(.)\1{2,}',r'\1\1',text)
    text = re.findall(r'[a-zA-Z]|\s',text)
    text = emoji.replace_emoji(text,'')
    text = re.sub(r'\s+',' ',text)
    tokens = word_tokenize(text)
    tokens = [token for token in tokens if token not in sw]
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(token) for token in tokens]
    return " ".join(tokens)

sentences = df_train['comment_text'].apply(clean_text)

vectorize_layer = tf.keras.layers.TextVectorization(
    max_tokens=100,
    output_mode='int'
)
vectorize_layer.adapt(sentences)
vectorized_text = vectorize_layer(sentences)
print(vectorized_text)
```
